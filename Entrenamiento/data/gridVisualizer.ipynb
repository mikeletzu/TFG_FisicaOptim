```python
import pandas as pd
import numpy as np

csv_path = 'clothDataset_107_.csv'
df = pd.read_csv(csv_path)

num_vertices = 25

selected_columns = []
for i in range(num_vertices):
    selected_columns.extend([f'y{i}', f'z{i}'])

df_filtered = df[selected_columns]

frame_0 = df_filtered.iloc[0].values
y_coords = frame_0[0::2]
z_coords = frame_0[1::2]

vertex_order = np.lexsort((z_coords, -y_coords))

print("El orden real de los vértices (row-major) es:")
print(repr(vertex_order))
```
